# Quantum Variational Classifier demo
This notebook shows how to train a simple quantum variational classifier (QNN) that learns the rule `y = 1 if x < 0.5 else 0`. The outputs below use SVG so GitHub's CODEX viewer can display them without binary warnings.

```python
%config InlineBackend.figure_formats = ['svg']
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

from app.dataset import DatasetConfig, generate_dataset
from app.qnn_model import QuantumVariationalClassifier, QNNConfig
from app.trainer import TrainerConfig, train_qnn
from app.plots import save_decision_boundary, save_loss_curve, EXAMPLES_DIR

# Generate data
x, y = generate_dataset(DatasetConfig(n_samples=60, noise=0.01, seed=5))

# Train a small QNN for a few epochs
model = QuantumVariationalClassifier(QNNConfig(seed=0))
model, losses = train_qnn(model, x, y, TrainerConfig(epochs=30, learning_rate=0.2))
preds = model.predict(x)

# Visualize decision boundary and loss. Files are SVG to stay text-based.
svg_boundary = save_decision_boundary(x, y, preds, EXAMPLES_DIR / 'decision_boundary_qnn.svg', 'QNN decision boundary')
svg_loss = save_loss_curve(losses, EXAMPLES_DIR / 'training_loss_qnn.svg')

svg_boundary, svg_loss
```

## Why this repo uses SVG instead of PNG (GitHub binary diff warning)
GitHub's CODEX viewer cannot preview binary files such as PNG or JPG and shows a warning saying "Binary files are not supported". SVG files are text-based, tiny, and render directly in the browser, so every plot in this project is stored as SVG.
